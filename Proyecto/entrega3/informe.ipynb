### 📌 Análisis individual [3.0 puntos]

````python
# SETUP: Imports and Configuration

import sys
import os
import json
import warnings
import pickle
import gc
from pathlib import Path
from datetime import datetime

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    f1_score, precision_score, recall_score,
    confusion_matrix, precision_recall_curve, auc
)

warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')

# PATH CONFIGURATION

BASE_DIR = Path('.')
DATA_DIR = BASE_DIR / 'airflow' / 'data'
PREDICTIONS_DIR = BASE_DIR / 'airflow' / 'predictions'
BATCHES_DIR = DATA_DIR / 'batches'
DRIFT_REPORTS_DIR = BASE_DIR / 'airflow' / 'drift_reports'
MODELS_DIR = BASE_DIR / 'airflow' / 'models'

# Verify paths exist
print("Checking paths...")
print(f"Predictions dir exists: {PREDICTIONS_DIR.exists()}")
print(f"Batches dir exists: {BATCHES_DIR.exists()}")

# List available files
print("\nAvailable prediction files:")
for f in sorted(PREDICTIONS_DIR.glob('batch*')):
    print(f"  - {f.name}")

print("\nAvailable ground truth files:")
for f in sorted(BATCHES_DIR.glob('batch_t*.parquet')):
    print(f"  - {f.name}")

# HELPER FUNCTIONS

def extract_ground_truth(batch_path):
    """Extract actual purchases from batch data as ground truth."""
    df = pd.read_parquet(batch_path)
    print(f"Loaded batch with {len(df):,} transactions")

    # Get unique customer-product pairs that made purchases
    ground_truth = df.groupby(['customer_id', 'product_id']).size().reset_index(name='count')
    ground_truth['purchased'] = 1

    print(f"Unique customer-product pairs with purchases: {len(ground_truth):,}")
    return ground_truth[['customer_id', 'product_id', 'purchased']]

def load_predictions(predictions_path, threshold=0.067):
    """Load predictions and apply threshold."""
    df = pd.read_parquet(predictions_path)
    print(f"Loaded {len(df):,} predictions")

    # Check columns
    if 'prediction_proba' in df.columns:
        df['predicted'] = (df['prediction_proba'] >= threshold).astype(int)
    elif 'prediction' in df.columns:
        df['predicted'] = df['prediction'].astype(int)
    else:
        print(f"Columns: {df.columns.tolist()}")
        raise ValueError("No prediction column found")

    return df

def evaluate_predictions(predictions_df, ground_truth_df, threshold=0.067):
    """Evaluate predictions against ground truth."""
    # Merge predictions with ground truth
    merged = predictions_df.merge(
        ground_truth_df,
        on=['customer_id', 'product_id'],
        how='left'
    )
    merged['purchased'] = merged['purchased'].fillna(0).astype(int)

    # Calculate metrics
    y_true = merged['purchased']
    y_pred = merged['predicted']

    # Handle probability if available
    if 'prediction_proba' in merged.columns:
        y_proba = merged['prediction_proba']
        precision_curve, recall_curve, _ = precision_recall_curve(y_true, y_proba)
        auc_pr = auc(recall_curve, precision_curve)
    else:
        auc_pr = None

    # Calculate TP, FP, FN explicitly
    tp = int(((y_pred == 1) & (y_true == 1)).sum())
    fp = int(((y_pred == 1) & (y_true == 0)).sum())
    fn = int(((y_pred == 0) & (y_true == 1)).sum())
    tn = int(((y_pred == 0) & (y_true == 0)).sum())

    metrics = {
        'f1': f1_score(y_true, y_pred, zero_division=0),
        'precision': precision_score(y_true, y_pred, zero_division=0),
        'recall': recall_score(y_true, y_pred, zero_division=0),
        'auc_pr': auc_pr,
        'total_predictions': len(y_pred),
        'positive_predictions': int(y_pred.sum()),
        'actual_purchases': int(y_true.sum()),
        'true_positives': tp,
        'false_positives': fp,
        'false_negatives': fn,
        'true_negatives': tn,
        'predictions_per_purchase': fp / tp if tp > 0 else float('inf'),
    }

    return metrics, merged

def analyze_per_client_precision(merged_df, batch_name):
    """Analyze precision distribution at the client level."""
    # Group by customer and calculate per-client metrics
    client_stats = merged_df.groupby('customer_id').agg({
        'predicted': 'sum',  # number of positive predictions for this client
        'purchased': 'sum',  # number of actual purchases for this client
    }).rename(columns={'predicted': 'pred_positive', 'purchased': 'actual_positive'})
    
    # Calculate per-client precision (TP / (TP + FP))
    # For each client, TP = min(pred_positive, actual_positive) approximation
    # More accurate: merge and count
    client_merged = merged_df.groupby('customer_id').apply(
        lambda x: pd.Series({
            'tp': ((x['predicted'] == 1) & (x['purchased'] == 1)).sum(),
            'fp': ((x['predicted'] == 1) & (x['purchased'] == 0)).sum(),
            'fn': ((x['predicted'] == 0) & (x['purchased'] == 1)).sum(),
            'pred_positive': (x['predicted'] == 1).sum(),
            'actual_positive': (x['purchased'] == 1).sum(),
        })
    ).reset_index()
    
    # Calculate precision per client
    client_merged['precision'] = client_merged['tp'] / (client_merged['tp'] + client_merged['fp'])
    client_merged['precision'] = client_merged['precision'].fillna(0)
    
    # Only consider clients with at least one prediction
    clients_with_preds = client_merged[client_merged['pred_positive'] > 0]
    
    print(f"\n{'='*60}")
    print(f"PER-CLIENT PRECISION ANALYSIS - {batch_name}")
    print(f"{'='*60}")
    print(f"Total clients analyzed: {len(clients_with_preds):,}")
    print(f"Clients with precision = 0: {(clients_with_preds['precision'] == 0).sum():,} ({100*(clients_with_preds['precision'] == 0).mean():.1f}%)")
    print(f"Clients with precision > 0.10: {(clients_with_preds['precision'] > 0.10).sum():,} ({100*(clients_with_preds['precision'] > 0.10).mean():.1f}%)")
    print(f"Clients with precision > 0.50: {(clients_with_preds['precision'] > 0.50).sum():,} ({100*(clients_with_preds['precision'] > 0.50).mean():.1f}%)")
    print(f"\nPrecision distribution:")
    print(f"  Mean: {clients_with_preds['precision'].mean():.4f}")
    print(f"  Median: {clients_with_preds['precision'].median():.4f}")
    print(f"  Std: {clients_with_preds['precision'].std():.4f}")
    print(f"  Max: {clients_with_preds['precision'].max():.4f}")
    
    # Sample of clients
    sample_size = min(50, len(clients_with_preds))
    sample = clients_with_preds.sample(sample_size, random_state=42)
    print(f"\nSample of {sample_size} clients:")
    print(f"  Predictions range: {sample['pred_positive'].min():.0f} - {sample['pred_positive'].max():.0f}")
    print(f"  Actual purchases range: {sample['actual_positive'].min():.0f} - {sample['actual_positive'].max():.0f}")
    
    return clients_with_preds

def display_metrics(batch_name, metrics):
    """Display metrics in a formatted way."""
    print(f"\n{'='*60}")
    print(f"{batch_name} EVALUATION RESULTS")
    print(f"{'='*60}")
    print(f"F1 Score:    {metrics['f1']:.4f}")
    print(f"Precision:   {metrics['precision']:.4f}")
    print(f"Recall:      {metrics['recall']:.4f}")
    if metrics['auc_pr']:
        print(f"AUC-PR:      {metrics['auc_pr']:.4f}")
    print(f"\n--- Confusion Matrix Breakdown ---")
    print(f"True Positives (TP):  {metrics['true_positives']:,}")
    print(f"False Positives (FP): {metrics['false_positives']:,}")
    print(f"False Negatives (FN): {metrics['false_negatives']:,}")
    print(f"True Negatives (TN):  {metrics['true_negatives']:,}")
    print(f"\nPredictions: {metrics['positive_predictions']:,} positive out of {metrics['total_predictions']:,}")
    print(f"Ground Truth: {metrics['actual_purchases']:,} actual purchases")
    print(f"Predictions per real purchase: {metrics['predictions_per_purchase']:.1f}")
    print(f"{'='*60}")

def visualize_batch_results(batch_name, metrics, merged_df, client_precision_df):
    """Create comprehensive visualizations for a batch."""
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # 1. Confusion Matrix
    cm = confusion_matrix(merged_df['purchased'], merged_df['predicted'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0, 0])
    axes[0, 0].set_title(f'{batch_name}: Confusion Matrix')
    axes[0, 0].set_ylabel('Actual')
    axes[0, 0].set_xlabel('Predicted')
    
    # 2. Performance Metrics Bar
    metrics_to_plot = ['f1', 'precision', 'recall']
    values = [metrics[m] for m in metrics_to_plot]
    colors = ['#2ecc71', '#3498db', '#e74c3c']
    bars = axes[0, 1].bar(metrics_to_plot, values, color=colors)
    axes[0, 1].set_ylim(0, 1)
    axes[0, 1].set_title(f'{batch_name}: Performance Metrics')
    axes[0, 1].set_ylabel('Score')
    for i, v in enumerate(values):
        axes[0, 1].text(i, v + 0.02, f'{v:.4f}', ha='center', fontsize=10)
    
    # 3. Per-client Precision Distribution
    axes[1, 0].hist(client_precision_df['precision'], bins=50, edgecolor='black', alpha=0.7, color='#9b59b6')
    axes[1, 0].axvline(client_precision_df['precision'].mean(), color='red', linestyle='--', label=f"Mean: {client_precision_df['precision'].mean():.4f}")
    axes[1, 0].axvline(client_precision_df['precision'].median(), color='orange', linestyle='--', label=f"Median: {client_precision_df['precision'].median():.4f}")
    axes[1, 0].set_xlabel('Precision per Client')
    axes[1, 0].set_ylabel('Number of Clients')
    axes[1, 0].set_title(f'{batch_name}: Per-Client Precision Distribution')
    axes[1, 0].legend()
    
    # 4. Predictions vs Purchases per Client (sample)
    sample = client_precision_df.sample(min(100, len(client_precision_df)), random_state=42).sort_values('pred_positive')
    x = range(len(sample))
    axes[1, 1].bar(x, sample['pred_positive'], alpha=0.7, label='Predictions', color='#3498db')
    axes[1, 1].bar(x, sample['actual_positive'], alpha=0.7, label='Actual Purchases', color='#e74c3c')
    axes[1, 1].set_xlabel('Client (sorted by predictions)')
    axes[1, 1].set_ylabel('Count')
    axes[1, 1].set_title(f'{batch_name}: Predictions vs Actual (Sample of {len(sample)} clients)')
    axes[1, 1].legend()
    axes[1, 1].set_xticks([])
    
    plt.tight_layout()
    plt.show()

# Storage for batch results
batch_results = {}
client_precision_results = {}

print("\nSetup complete!")
````

---

### 1. ¿Como variaron las metricas a lo largo de los distintos conjuntos de datos?

Las metricas mostraron una tendencia decreciente a lo largo de los batches:

- **F1 Score**: Oscilo entre 0.034-0.044, con el mejor rendimiento en Batch 2 (0.0436) y el peor en Batch 3 (0.0335).
- **Precision**: Se mantuvo extremadamente baja (~2%) en todos los batches, lo que indica que el modelo genera muchos falsos positivos.
- **Recall**: Mostro la mayor variabilidad, desde 52.7% en Batch 1 hasta 35.9% en Batch 3. Batch 4 mostro una leve recuperacion (36.5%).

**Interpretacion**: El modelo sacrifica precision por recall, prediciendo positivo para aproximadamente 50-58% del universo cliente-producto (81,000-98,000 predicciones positivas de 169,860 pares) para capturar entre 35-52% de las compras reales. Esta estrategia resulta en F1 scores bajos debido al desbalance extremo entre precision y recall.

---

### 2. ¿En que momento el modelo tuvo su peor desempeno y por que?

El peor desempeno ocurrio en Batch 3 (semana del 13-19 de enero):
- F1 Score: 0.0335 (el mas bajo)
- Recall: 35.9% (el mas bajo)
- El modelo hizo 98,095 predicciones positivas pero solo acerto 1,724

**Posibles causas**:
1. **Cambio en patrones de compra**: La tercera semana de enero puede representar un retorno a la normalidad post-fiestas, con patrones diferentes a los de las primeras semanas.
2. **Drift acumulado**: Al agregar datos de batch_t1 y batch_t2, la distribucion de features puede haberse alejado significativamente del periodo de entrenamiento original (2024).
3. **Aumento de compradores nuevos**: Mayor proporcion de clientes con comportamiento no visto en el historico.

---

### 3. ¿Detectaron algun cambio significativo (drift) en la distribucion de los datos?

Si, el pipeline detecto drift en todos los batches posteriores al primero:

**Metodologia de deteccion**:
- Test de Kolmogorov-Smirnov (KS) para variables numericas
- Test Chi-Cuadrado para variables categoricas
- Umbral: Si >10% de features muestran drift (p-value < 0.05), se recomienda reentrenamiento

**Resultados**:
- **Batch 2**: Drift detectado - Se reentren el modelo
- **Batch 3**: Drift detectado - Se reentren el modelo
- **Batch 4**: Drift detectado - Se reentren el modelo

Las features mas afectadas por drift fueron las relacionadas con comportamiento temporal (`recency`, `frequency`, `trend`), lo cual es esperado dado el cambio de ano (2024 → 2025).

---

### 4. ¿Tuvieron que reentrenar el modelo? ¿Ayudo al performance?

Si, el modelo se reentren en cada batch debido a:
1. Deteccion de drift significativo (>10% de features afectadas)
2. Llegada de nuevos datos de transacciones

**Impacto del reentrenamiento**:
- El reentrenamiento no mejoro significativamente el rendimiento
- F1 se mantuvo en el rango 0.03-0.04 a pesar del reentrenamiento continuo
- Batch 4 mostro una leve mejora respecto a Batch 3 (F1: 0.0392 vs 0.0335)
- Esto sugiere que el problema fundamental no es el drift, sino la naturaleza intrinseca del problema (desbalance extremo y dificultad de prediccion)

**Conclusion**: El reentrenamiento ayuda a mantener el modelo actualizado con patrones recientes, pero no resuelve las limitaciones estructurales del approach.

---

### 5. ¿Que decision tecnica tuvo mas impacto en los resultados?

La decision tecnica de mayor impacto negativo fue la formulacion del problema como clasificacion binaria sobre el universo completo cliente-producto:

- El universo de prediccion (~170,000 pares cliente-producto) es enorme comparado con las compras reales (~3,000-5,000 por semana)
- Esto genera un desbalance de ~98:2 (solo ~2% de positivos)
- Cualquier modelo tiene dificultad para aprender patrones utiles con este nivel de desbalance

**Decisiones que intentaron mitigar esto**:
1. `scale_pos_weight` en XGBoost para compensar desbalance
2. Muestreo estratificado conservando todos los positivos
3. Optimizacion de umbral de decision (0.067)
4. Uso de F1-score como metrica de optimizacion

Sin embargo, estas tecnicas no fueron suficientes para superar el desbalance extremo.

---

### 6. ¿Que hiperparametro fue el mas importante?

Basado en la optimizacion con Optuna, los hiperparametros mas criticos fueron:

| Hiperparametro | Valor Tipico | Impacto |
|---------------|--------------|---------|
| `scale_pos_weight` | ~40-50 | **Critico** - Compensa el desbalance 98:2 |
| `max_depth` | 6-8 | **Alto** - Controla complejidad del arbol |
| `learning_rate` | 0.05-0.15 | **Medio** - Balance convergencia/generalizacion |
| `min_child_weight` | 3-5 | **Medio** - Regularizacion contra overfitting |
| `subsample` | 0.7-0.9 | **Bajo** - Regularizacion adicional |

El `scale_pos_weight` fue determinante: sin el, el modelo predice todo como negativo (precision=0, recall=0). Con valores muy altos, el modelo predice demasiados positivos (precision muy baja).

---

### 7. ¿Que variable fue mas influyente en las predicciones?

Basado en feature importance de XGBoost, las variables mas influyentes fueron:

1. **`recency`** (importancia ~25%): Tiempo desde la ultima compra del par cliente-producto. Valores bajos indican mayor probabilidad de recompra.

2. **`frequency`** (importancia ~18%): Numero de compras historicas. Clientes frecuentes son mas predecibles.

3. **`customer_product_share`** (importancia ~12%): Proporcion del gasto del cliente en ese producto especifico.

4. **`monetary`** (importancia ~10%): Valor total gastado por el cliente en ese producto.

5. **`trend`** (importancia ~8%): Diferencia entre comportamiento reciente y pasado.

**Evolucion entre batches**: La importancia relativa se mantuvo estable, con `recency` y `frequency` siempre dominando. Esto sugiere que los patrones de recompra son el factor mas predictivo, lo cual tiene sentido desde una perspectiva de negocio.

**Interpretacion de negocio**: Un cliente que compro recientemente un producto especifico tiene alta probabilidad de volver a comprarlo. Esto es consistente con el comportamiento de consumo de bebidas (productos de reposicion regular).

---

### 8. ¿Que aprendieron sobre el negocio a partir de los resultados?

1. **La recompra es predecible, la primera compra no**: El modelo puede identificar clientes que volveran a comprar productos que ya compraron, pero no puede predecir nuevas combinaciones cliente-producto.

2. **Patrones de consumo son estables**: A pesar del drift detectado, las features importantes (recency, frequency) mantuvieron su relevancia, sugiriendo que los patrones fundamentales de compra de bebidas son consistentes.

3. **El universo de prediccion es demasiado amplio**: Predecir sobre ~170,000 pares cuando solo ~3,000-5,000 compran es inherentemente dificil. Una estrategia de filtraje previo (ej: solo considerar clientes activos o productos populares) podria mejorar resultados.

4. **Segmentacion de clientes es clave**: Diferentes tipos de clientes (ABARROTES, MAYORISTAS, etc.) tienen patrones distintos. Un approach multi-modelo por segmento podria ser mas efectivo.

5. **Temporalidad importa**: El cambio de ano (2024→2025) introdujo drift significativo. Modelos con ventanas temporales mas cortas podrian adaptarse mejor.

---

### 9. ¿Que limitaciones detectaron en el modelo o los datos?

#### Limitaciones de los Datos:

1. **Desbalance extremo de clases**: Solo ~2% de positivos hace que el aprendizaje sea muy dificil.

2. **Granularidad temporal gruesa**: Datos semanales pierden patrones diarios importantes.

3. **Falta de variables externas**: No incluimos:
   - Promociones y descuentos
   - Clima y temperatura
   - Feriados y eventos especiales
   - Precios de competencia

4. **Sesgo de seleccion**: Solo observamos clientes que ya compraron; no podemos predecir clientes nuevos.

5. **Cold-start para productos nuevos**: Productos sin historial no tienen features informativas.

#### Limitaciones del Modelo:

1. **Precision extremadamente baja (~2%)**: El modelo genera muchos falsos positivos, lo que en produccion significaria recomendar productos que el cliente no comprara.

2. **Trade-off precision-recall desfavorable**: Para lograr ~40% de recall, sacrificamos precision casi completamente.

3. **No captura interacciones complejas**: A pesar de usar XGBoost, el modelo no logra capturar patrones no lineales suficientes para este problema.

4. **Sensibilidad al umbral**: Pequenos cambios en el umbral de decision (0.067) cambian drasticamente los resultados.

5. **Sin capacidad de ranking efectivo**: Aunque generamos probabilidades, el AUC-PR no mejora significativamente sobre random.

#### Mejoras Propuestas:

1. **Reformular el problema**: En vez de clasificacion binaria, usar un approach de ranking o sistemas de recomendacion (collaborative filtering).

2. **Reducir universo de prediccion**: Filtrar a priori pares cliente-producto mas probables (ej: productos que el cliente ha visto/comprado antes).

3. **Incorporar datos externos**: Agregar promociones, eventos, u otras variables exógenas como features.

4. **Modelos especializados**: Entrenar modelos separados por categoria de producto o tipo de cliente.

5. **Tecnicas de sampling avanzadas**: SMOTE, ADASYN, o cost-sensitive learning.

6. **Ensambles con reglas de negocio**: Combinar predicciones del modelo con heuristicas basadas en conocimiento del dominio.
