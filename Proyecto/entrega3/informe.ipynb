### 📌 Análisis individual [3.0 puntos]

````python
# SETUP: Imports and Configuration

import sys
import os
import json
import warnings
import pickle
import gc
from pathlib import Path
from datetime import datetime

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    f1_score, precision_score, recall_score,
    confusion_matrix, precision_recall_curve, auc
)

warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')

# PATH CONFIGURATION

BASE_DIR = Path('.')
DATA_DIR = BASE_DIR / 'airflow' / 'data'
PREDICTIONS_DIR = BASE_DIR / 'airflow' / 'predictions'
BATCHES_DIR = DATA_DIR / 'batches'
DRIFT_REPORTS_DIR = BASE_DIR / 'airflow' / 'drift_reports'
MODELS_DIR = BASE_DIR / 'airflow' / 'models'

# Verify paths exist
print("Checking paths...")
print(f"Predictions dir exists: {PREDICTIONS_DIR.exists()}")
print(f"Batches dir exists: {BATCHES_DIR.exists()}")

print("\nAvailable prediction files:")
for f in sorted(PREDICTIONS_DIR.glob('batch*')):
    print(f"  - {f.name}")

print("\nAvailable ground truth files:")
for f in sorted(BATCHES_DIR.glob('batch_t*.parquet')):
    print(f"  - {f.name}")

def extract_ground_truth(batch_path):
    """Extract actual purchases from batch data as ground truth."""
    df = pd.read_parquet(batch_path)
    print(f"Loaded batch with {len(df):,} transactions")

    # Get unique customer-product pairs that made purchases
    ground_truth = df.groupby(['customer_id', 'product_id']).size().reset_index(name='count')
    ground_truth['purchased'] = 1

    print(f"Unique customer-product pairs with purchases: {len(ground_truth):,}")
    return ground_truth[['customer_id', 'product_id', 'purchased']]

def load_predictions(predictions_path, threshold=0.067):
    """Load predictions and apply threshold."""
    df = pd.read_parquet(predictions_path)
    print(f"Loaded {len(df):,} predictions")

    # Check columns
    if 'prediction_proba' in df.columns:
        df['predicted'] = (df['prediction_proba'] >= threshold).astype(int)
    elif 'prediction' in df.columns:
        df['predicted'] = df['prediction'].astype(int)
    else:
        print(f"Columns: {df.columns.tolist()}")
        raise ValueError("No prediction column found")

    return df

def evaluate_predictions(predictions_df, ground_truth_df, threshold=0.067):
    """Evaluate predictions against ground truth."""
    # Merge predictions with ground truth
    merged = predictions_df.merge(
        ground_truth_df,
        on=['customer_id', 'product_id'],
        how='left'
    )
    merged['purchased'] = merged['purchased'].fillna(0).astype(int)

    # Calculate metrics
    y_true = merged['purchased']
    y_pred = merged['predicted']

    # Handle probability if available
    if 'prediction_proba' in merged.columns:
        y_proba = merged['prediction_proba']
        precision_curve, recall_curve, _ = precision_recall_curve(y_true, y_proba)
        auc_pr = auc(recall_curve, precision_curve)
    else:
        auc_pr = None

    # Calculate TP, FP, FN explicitly
    tp = int(((y_pred == 1) & (y_true == 1)).sum())
    fp = int(((y_pred == 1) & (y_true == 0)).sum())
    fn = int(((y_pred == 0) & (y_true == 1)).sum())
    tn = int(((y_pred == 0) & (y_true == 0)).sum())

    metrics = {
        'f1': f1_score(y_true, y_pred, zero_division=0),
        'precision': precision_score(y_true, y_pred, zero_division=0),
        'recall': recall_score(y_true, y_pred, zero_division=0),
        'auc_pr': auc_pr,
        'total_predictions': len(y_pred),
        'positive_predictions': int(y_pred.sum()),
        'actual_purchases': int(y_true.sum()),
        'true_positives': tp,
        'false_positives': fp,
        'false_negatives': fn,
        'true_negatives': tn,
        'predictions_per_purchase': fp / tp if tp > 0 else float('inf'),
    }

    return metrics, merged

def analyze_per_client_precision(merged_df, batch_name):
    """Analyze precision distribution at the client level."""
    # Group by customer and calculate per-client metrics
    client_stats = merged_df.groupby('customer_id').agg({
        'predicted': 'sum',  # number of positive predictions for this client
        'purchased': 'sum',  # number of actual purchases for this client
    }).rename(columns={'predicted': 'pred_positive', 'purchased': 'actual_positive'})
    
    # Calculate per-client precision (TP / (TP + FP))
    # For each client, TP = min(pred_positive, actual_positive) approximation
    # More accurate: merge and count
    client_merged = merged_df.groupby('customer_id').apply(
        lambda x: pd.Series({
            'tp': ((x['predicted'] == 1) & (x['purchased'] == 1)).sum(),
            'fp': ((x['predicted'] == 1) & (x['purchased'] == 0)).sum(),
            'fn': ((x['predicted'] == 0) & (x['purchased'] == 1)).sum(),
            'pred_positive': (x['predicted'] == 1).sum(),
            'actual_positive': (x['purchased'] == 1).sum(),
        })
    ).reset_index()
    
    # Calculate precision per client
    client_merged['precision'] = client_merged['tp'] / (client_merged['tp'] + client_merged['fp'])
    client_merged['precision'] = client_merged['precision'].fillna(0)
    
    # Only consider clients with at least one prediction
    clients_with_preds = client_merged[client_merged['pred_positive'] > 0]
    
    print(f"\n{'='*60}")
    print(f"PER-CLIENT PRECISION ANALYSIS - {batch_name}")
    print(f"{'='*60}")
    print(f"Total clients analyzed: {len(clients_with_preds):,}")
    print(f"Clients with precision = 0: {(clients_with_preds['precision'] == 0).sum():,} ({100*(clients_with_preds['precision'] == 0).mean():.1f}%)")
    print(f"Clients with precision > 0.10: {(clients_with_preds['precision'] > 0.10).sum():,} ({100*(clients_with_preds['precision'] > 0.10).mean():.1f}%)")
    print(f"Clients with precision > 0.50: {(clients_with_preds['precision'] > 0.50).sum():,} ({100*(clients_with_preds['precision'] > 0.50).mean():.1f}%)")
    print(f"\nPrecision distribution:")
    print(f"  Mean: {clients_with_preds['precision'].mean():.4f}")
    print(f"  Median: {clients_with_preds['precision'].median():.4f}")
    print(f"  Std: {clients_with_preds['precision'].std():.4f}")
    print(f"  Max: {clients_with_preds['precision'].max():.4f}")
    
    # Sample of clients
    sample_size = min(50, len(clients_with_preds))
    sample = clients_with_preds.sample(sample_size, random_state=42)
    print(f"\nSample of {sample_size} clients:")
    print(f"  Predictions range: {sample['pred_positive'].min():.0f} - {sample['pred_positive'].max():.0f}")
    print(f"  Actual purchases range: {sample['actual_positive'].min():.0f} - {sample['actual_positive'].max():.0f}")
    
    return clients_with_preds

def display_metrics(batch_name, metrics):
    """Display metrics in a formatted way."""
    print(f"\n{'='*60}")
    print(f"{batch_name} EVALUATION RESULTS")
    print(f"{'='*60}")
    print(f"F1 Score:    {metrics['f1']:.4f}")
    print(f"Precision:   {metrics['precision']:.4f}")
    print(f"Recall:      {metrics['recall']:.4f}")
    if metrics['auc_pr']:
        print(f"AUC-PR:      {metrics['auc_pr']:.4f}")
    print(f"\n--- Confusion Matrix Breakdown ---")
    print(f"True Positives (TP):  {metrics['true_positives']:,}")
    print(f"False Positives (FP): {metrics['false_positives']:,}")
    print(f"False Negatives (FN): {metrics['false_negatives']:,}")
    print(f"True Negatives (TN):  {metrics['true_negatives']:,}")
    print(f"\nPredictions: {metrics['positive_predictions']:,} positive out of {metrics['total_predictions']:,}")
    print(f"Ground Truth: {metrics['actual_purchases']:,} actual purchases")
    print(f"Predictions per real purchase: {metrics['predictions_per_purchase']:.1f}")
    print(f"{'='*60}")

def visualize_batch_results(batch_name, metrics, merged_df, client_precision_df):
    """Create comprehensive visualizations for a batch."""
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # 1. Confusion Matrix
    cm = confusion_matrix(merged_df['purchased'], merged_df['predicted'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0, 0])
    axes[0, 0].set_title(f'{batch_name}: Confusion Matrix')
    axes[0, 0].set_ylabel('Actual')
    axes[0, 0].set_xlabel('Predicted')
    
    # 2. Performance Metrics Bar
    metrics_to_plot = ['f1', 'precision', 'recall']
    values = [metrics[m] for m in metrics_to_plot]
    colors = ['#2ecc71', '#3498db', '#e74c3c']
    bars = axes[0, 1].bar(metrics_to_plot, values, color=colors)
    axes[0, 1].set_ylim(0, 1)
    axes[0, 1].set_title(f'{batch_name}: Performance Metrics')
    axes[0, 1].set_ylabel('Score')
    for i, v in enumerate(values):
        axes[0, 1].text(i, v + 0.02, f'{v:.4f}', ha='center', fontsize=10)
    
    # 3. Per-client Precision Distribution
    axes[1, 0].hist(client_precision_df['precision'], bins=50, edgecolor='black', alpha=0.7, color='#9b59b6')
    axes[1, 0].axvline(client_precision_df['precision'].mean(), color='red', linestyle='--', label=f"Mean: {client_precision_df['precision'].mean():.4f}")
    axes[1, 0].axvline(client_precision_df['precision'].median(), color='orange', linestyle='--', label=f"Median: {client_precision_df['precision'].median():.4f}")
    axes[1, 0].set_xlabel('Precision per Client')
    axes[1, 0].set_ylabel('Number of Clients')
    axes[1, 0].set_title(f'{batch_name}: Per-Client Precision Distribution')
    axes[1, 0].legend()
    
    # 4. Predictions vs Purchases per Client (sample)
    sample = client_precision_df.sample(min(100, len(client_precision_df)), random_state=42).sort_values('pred_positive')
    x = range(len(sample))
    axes[1, 1].bar(x, sample['pred_positive'], alpha=0.7, label='Predictions', color='#3498db')
    axes[1, 1].bar(x, sample['actual_positive'], alpha=0.7, label='Actual Purchases', color='#e74c3c')
    axes[1, 1].set_xlabel('Client (sorted by predictions)')
    axes[1, 1].set_ylabel('Count')
    axes[1, 1].set_title(f'{batch_name}: Predictions vs Actual (Sample of {len(sample)} clients)')
    axes[1, 1].legend()
    axes[1, 1].set_xticks([])
    
    plt.tight_layout()
    plt.show()

# Storage for batch results
batch_results = {}
client_precision_results = {}

print("\nSetup complete!")
````

**Contexto**: Este es el primer batch de evaluacion. El modelo fue entrenado desde cero usando datos historicos de 2024 (transacciones originales). No hay drift previo ya que es la primera ejecucion.

### 📌 Conclusiones y Aprendizajes [1.0 puntos]

### Conclusiones

**1. ¿Como variaron las metricas a lo largo de los distintos conjuntos de datos?**

El F1 oscilo entre 0.033 y 0.044 - variacion pequena pero consistentemente baja. El recall fue lo que mas cambio: empezo en 52.7% (Batch 1) y bajo a ~36% en los siguientes. La precision se mantuvo fija en ~2% en todos los batches. Basicamente, el modelo empezo "agresivo" (muchas predicciones, buen recall) y se fue volviendo mas conservador sin mejorar la precision.

**2. ¿En que momento el modelo tuvo su peor desempeño y por que?**

Batch 3 fue el peor (F1: 0.0335). No tenemos una causa clara - las features importantes no cambiaron, el drift detectado fue similar a otros batches. Puede ser simplemente variabilidad en los datos de esa semana particular, o que el drift acumulado de 3 semanas finalmente afecto al modelo.

**3. ¿Detectaron algun cambio significativo (drift) en la distribucion de los datos?**

Si, en todos los batches. Usamos tests KS para variables numericas y Chi-cuadrado para categoricas, con umbral de 10% de features afectadas. Probablemente fuimos muy sensibles - 10% es bajo y termino forzando reentrenamiento cada semana. Las features mas afectadas fueron las temporales (`recency`, `trend`), lo cual tiene sentido por el cambio de año.

**4. ¿Tuvieron que reentrenar el modelo? ¿Ayudo al performance?**

Reentrenamos en cada batch por el drift detectado. ¿Ayudo? No realmente. Las metricas no mejoraron de forma consistente - Batch 4 fue un poco mejor que Batch 3, pero la diferencia es minima. El modelo parece tener un techo que el reentrenamiento no logra superar.

**5. ¿Que decision tecnica tuvo mas impacto en los resultados?**

La formulacion del problema como clasificacion binaria sobre todos los pares cliente-producto. Tenemos ~170,000 pares y solo ~3,000-5,000 compras reales por semana (desbalance 98:2). Por mas que ajustemos `scale_pos_weight` y optimicemos umbrales, un clasificador binario lo tiene muy dificil con esos numeros. Debimos filtrar el universo de prediccion o usar otro approach (recomendacion, ranking).

**6. ¿Que hiperparametro fue el mas importante?**

`scale_pos_weight` (~45). Sin el, el modelo predice todo negativo. Con valores muy altos, predice demasiados positivos. Es el unico hiperparametro que realmente cambia el comportamiento del modelo de forma drastica para este problema desbalanceado. `max_depth` y `learning_rate` tambien importan, pero en menor medida.

**7. ¿Que variable fue mas influyente? ¿Como cambio su importancia entre batches?**

`recency` domino en todos los batches (SHAP ~0.15-0.20), seguido de `frequency` y `customer_product_share`. El ranking se mantuvo estable - el reentrenamiento no "descubrio" patrones nuevos, solo actualizo los mismos. Esto sugiere que el comportamiento de compra de bebidas es predecible pero nuestro modelo no logra capitalizarlo bien.

**8. ¿Que aprendieron sobre el negocio?**

Los clientes de bebidas son criaturas de habito - compran lo mismo repetidamente. Las features RFM (recency, frequency, monetary) capturan bien esto. Pero hay mucha variabilidad individual: clientes con "buenas" features que no compran, y viceversa. Quizas variables externas (promociones, clima, feriados) explicarian parte de esa variabilidad.

**9. ¿Que limitaciones detectaron?**

- **Desbalance extremo**: 98% negativos hace el aprendizaje muy dificil
- **Precision inutilizable**: 2% significa que 98 de 100 recomendaciones serian erroneas en produccion
- **Sin variables externas**: no tenemos promociones, precios, eventos
- **Cold-start**: productos/clientes nuevos no tienen features informativas
- **Universo muy amplio**: predecir sobre 170,000 pares cuando solo 3,000 compran es un problema mal planteado

El pipeline funciona bien (Airflow, MLflow, drift detection, Optuna), pero el problema de negocio requeria otro approach desde el inicio. Un sistema de recomendacion o al menos filtrar a pares con historial reciente habria sido mas sensato.
